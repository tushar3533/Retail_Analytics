```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("-", "_")
)

print(df.columns.tolist())
```

```python
df = df.drop_duplicates()
```

```python
df["order_date"] = pd.to_datetime(
    df["order_date"], errors="coerce"
)

df["ship_date"] = pd.to_datetime(
    df["ship_date"], errors="coerce"
)
```

```python
df=df.drop(columns=["row_id"])   #row_id is not an essential or usefull column
```

```python
print(df[["sales", "profit", "quantity", "discount"]].describe())

print("Invalid dates:", df["order_date"].isna().sum())

print("Ship dates before order dates:",
      (df["ship_date"] < df["order_date"]).sum())
```

```python
df["delivery_days"] = (
    df["ship_date"] - df["order_date"]
).dt.days
```

```python
df["order_year"] = df["order_date"].dt.year
```

```python
df["order_month"] = df["order_date"].dt.month
```

```python
df["month_name"] = df["order_date"].dt.strftime("%B")
```

```python
df["profit_margin"] = np.where(
    df["sales"] != 0,
    (df["profit"] / df["sales"]) * 100,
    0
)
```

```python
df["discount_category"] = pd.cut(
    df["discount"],
    bins=[-0.001, 0, 0.1, 0.2, 1],
    labels=["No Discount", "Low", "Medium", "High"]
)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load cleaned dataset
file_path = r"C:\Users\tusha\OneDrive\Desktop\Retail_Analytics\Dataset\superstore_clean.csv"

df = pd.read_csv(
    file_path,
    parse_dates=["order_date", "ship_date"]
)

# Verify successful loading
print("Cleaned dataset loaded successfully!")

print("Dataset Shape:", df.shape)

display(df.head())

df.info()
```

```python
total_sales = df["sales"].sum()

total_profit = df["profit"].sum()

total_orders = df["order_id"].nunique()

total_customers = df["customer_id"].nunique()

total_quantity = df["quantity"].sum()

average_order_value = total_sales / total_orders

profit_margin = (
    total_profit / total_sales
) * 100

print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Orders:", total_orders)
print("Total Customers:", total_customers)
print("Total Quantity:", total_quantity)
print("Average Order Value:", round(average_order_value, 2))
print("Profit Margin:", round(profit_margin, 2), "%")
```

```python
#Category generating more revenue
category_sales = (
    df.groupby("category")["sales"]
    .sum()
    .sort_values(ascending=False)
)

print(category_sales)
```

```python
category_sales.plot(
    kind="bar",
    title="Sales by Category",
    figsize=(8, 5)
)

plt.xlabel("Category")
plt.ylabel("Sales")
plt.tight_layout()
plt.show()
```

```python
#Category wise profits
category_profit = (
    df.groupby("category")["profit"]
    .sum()
)

print(category_profit)
```

```python
category_profit.plot(
    kind="bar",
    title="Profit by Category",
    figsize=(8, 5)
)

plt.xlabel("Category")
plt.ylabel("Profit")
plt.show()
```

```python
#Monthly sale trends
monthly_sales = (
    df.groupby("order_month")["sales"]
    .sum()
)

monthly_sales.plot(
    kind="line",
    marker="o",
    figsize=(10, 5),
    title="Monthly Sales Trend"
)

plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(range(1, 13))
plt.grid()
plt.show()
```

```python
region_analysis = (
    df.groupby("region")[["sales", "profit"]]
    .sum()
    .sort_values("sales", ascending=False)
)

print(region_analysis)
```

```python
top_products = (
    df.groupby("product_name")["sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(top_products)
```

```python
customer_analysis = (
    df.groupby("segment")
    .agg(
        total_sales=("sales", "sum"),
        total_profit=("profit", "sum"),
        total_customers=("customer_id", "nunique")
    )
)

print(customer_analysis)
```
